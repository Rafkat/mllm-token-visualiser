```python
from transformers import AutoProcessor

model_path = "Qwen/Qwen3-VL-8B-Instruct"

processor = AutoProcessor.from_pretrained(model_path, trust_remote_code=True)

messages = [
    {
        'role': 'user',
        'content': [
            {
                "type": "text",
                "text": "Hey there!"
            }
        ]
    }
]

inputs = processor.apply_chat_template(
    messages,
    return_tensors="pt",
    tokenize=True,
    add_generation_prompt=False,
    return_dict=True
)

total_tokens = inputs["input_ids"][0].shape[0]
```

```python
import torch
special_ids_tensor = torch.tensor(processor.tokenizer.all_special_ids)
mask = ~torch.isin(inputs['input_ids'][0], special_ids_tensor)
inputs_without_special_tokens = inputs['input_ids'][0][mask]
inputs_without_special_tokens
```
